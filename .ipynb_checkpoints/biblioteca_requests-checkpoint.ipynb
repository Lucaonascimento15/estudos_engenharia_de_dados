```python
## - requests: coletar dados de uma API pública.
```

```python
import requests
#GET - Pegar uma informação
#requisicao = requests.get("https://economia.awesomeapi.com.br/last/USD-BRL,EUR-BRL,BTC-BRL")
requisicao = requests.get("https://teste-9e179-default-rtdb.firebaseio.com/.json")
print(requisicao)

print (requisicao.json())
```

```python
#POST - postar,criar, imputar uma informação

informacao = '{"Nome": "Cabeça de porco"}'

requisicao = requests.post("https://teste-9e179-default-rtdb.firebaseio.com/.json", data=informacao)
print(requisicao)
print (requisicao.json())
```

```python
#PATCH - atualizar uma informação

informacao = '{"Nome": "Priscila", "Idade": "19", "Endereco": "Rua Cais Das Artes"}'

requisicao = requests.patch("https://teste-9e179-default-rtdb.firebaseio.com/-P5xkhtevDRZJWB21N_A/.json", data=informacao)
print(requisicao)
print (requisicao.json())
```

```python
#DELETE - Deletar uma informação
requisicao = requests.delete("https://teste-9e179-default-rtdb.firebaseio.com/-P5xtlTavL9cKI0_Q-S0/.json")
print(requisicao)
print (requisicao.json())
```
